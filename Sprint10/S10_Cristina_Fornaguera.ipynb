1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

+ Llista de publicacions (/posts)
+ Llista d’usuaris (/users)
+ Llista de tasques (/todos)

```python
import requests
from requests.exceptions import HTTPError
def get_json(url):
    try:
        resp = requests.get(url)
        resp.raise_for_status() #  HTTPError for status codes between 400 and 600
    except HTTPError as http_err:
        print(f"HTTP error occurred: {http_err}\n")
    except Exception as err:
        print(f"Other error occurred: {err}\n")
    else:
        return resp, resp.json()
```

```python
# /posts
posts, posts_data = get_json("https://jsonplaceholder.typicode.com/posts")
posts_data
```

```python
# /users
users, users_data = get_json("https://jsonplaceholder.typicode.com/users")
users_data
```

```python
# /todos
todos, todos_data = get_json("https://jsonplaceholder.typicode.com/todos")
todos_data
```

```python
# Cantidades de cada recurso:
print(f"Cantidad de posts: {len(posts_data)}")
print(f"Cantidad de users: {len(users_data)}")
print(f"Cantidad de todos: {len(todos_data)}")

# Códigos de estado de cada petición:
print(f"Estado de posts: {posts.status_code}")
print(f"Estado de users: {users.status_code}")
print(f"Estado de todos: {todos.status_code}")
```

3. Fes una petició GET senzilla:

Mostra el codi d'estat de la resposta.
Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
import requests
url = "https://openlibrary.org/search.json"
params = {
    "title": "Inteligencia artificial",
    "limit": 5  # Limita los resultados para no saturar el servicio
}

response = requests.get(url, params=params)
ol_json = response.json()
content_type = response.headers.get("Content-Type", "")
# mostrar el tipo de contenido y el código de estado de la respuesta
print(f"Content-Type: {content_type}")
print(f"Status Code: {response.status_code}")
# mostrar el contenido de la respuesta en formato JSON
if response.status_code == 200:
    data = response.json()
    libros = data.get("docs", [])

    print(f"Total de resultados encontrados: {data.get('numFound')}\n")

    # Recorrer los 5 primeros resultados (si se tuvieran más de 5) y mostrar título, autor y año de publicación
    for libro in libros[:5]:
        titulo = libro.get("title", "Sin título")
        autores = ", ".join(libro.get("author_name", ["Autor desconocido"]))
        anio = libro.get("first_publish_year", "N/A")

        print(f" Título: {titulo}")
        print(f" Autor(es): {autores}")
        print(f" Año de publicación: {anio}")
        print("-" * 40)
else:
    print(f"Error al realizar la búsqueda: Status {response.status_code}")
```

4. Converteix les dades a un DataFrame de pandas:

Mostra les primeres files del DataFrame.

```python
import pandas as pd
# Crear un DataFrame a partir de los resultados
df_libros = pd.DataFrame(libros)
display(df_libros.head())  # Mostrar las primeras filas del DataFrame
```

Nivell 3  
API d'Open Data Barcelona  
1. Utilitza l’API d’Open Data BCN per:  
Open Data BCN  
https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS  
Cercar un dataset del teu interès mitjançant package_search i package_show.

```python
import requests

BASE_URL = "https://opendata-ajuntament.barcelona.cat/data/api/action/"

# Uso de 'package_search' para encontrar dataset de defunciones
search_url = f"{BASE_URL}package_search"
search_params = {
    "q": "defuncions",
    "rows": 5
}

response_search = requests.get(search_url, params=search_params)

if response_search.status_code == 200:
    search_data = response_search.json()

    if search_data.get("success"):
        results = search_data["result"]["results"]
        print(f"S'han trobat {search_data['result']['count']} datasets relacionats.\n")

        # Guardamos el ID del primer dataset
        dataset_id = results[0]["name"]
        print(f" Seleccionat dataset: {results[0]['title']} (ID/Name: {dataset_id})\n")
    else:
        print("La cerca no ha tingut èxit.")
else:
    print(f"Error en package_search: Status {response_search.status_code}")

# Uso de 'package_show' para obtener todos los detalles del dataset
show_url = f"{BASE_URL}package_show"
show_params = {
    "id": dataset_id  # Usamos el ID obtenido en la búsqueda
}

response_show = requests.get(show_url, params=show_params)

if response_show.status_code == 200:
    show_data = response_show.json()

    if show_data.get("success"):
        dataset_detail = show_data["result"]

        print(" DETALLS DEL DATASET (package_show):")
        print(f"• Títol: {dataset_detail.get('title')}")
        print(f"• Llicència: {dataset_detail.get('license_title')}")
        print(f"• Nombre de recursos disponibles: {len(dataset_detail.get('resources', []))}\n")

        print(" RECURSOS (Fitxers de dades disponibles):")
        for res in dataset_detail.get("resources", [])[:5]:  # Limitamos a mostrar solo los primeros 5 recursos
            print(f"  - Nom: {res.get('name')}")
            print(f"    Format: {res.get('format')}")
            print(f"    Resource ID: {res.get('id')}")
            print(f"    URL descàrrega: {res.get('url')}")
            print("-" * 50)
else:
    print(f"Error en package_show: Status {response_show.status_code}")
```

2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
import requests

BASE_URL = "https://opendata-ajuntament.barcelona.cat/data/api/action/"

# ID del recurso CSV/JSON seleccionado del dataset de defunciones
RESOURCE_ID = "840f6761-46c5-4bf1-93e5-906bec0664bc"  # ID de ejemplo de un recurso ndel catálogo
```

3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

```python
datastore_url = f"{BASE_URL}datastore_search"
params = {
    "resource_id": RESOURCE_ID,
    "limit": 100  # Pedimos 100 registros
}

response = requests.get(datastore_url, params=params)

if response.status_code == 200:
    data = response.json()

    if data.get("success"):
        result = data["result"]
        records = result.get("records", [])
        fields = [field["id"] for field in result.get("fields", [])]

        print(" RECURS SELECCIONAT AMB ÈXIT")
        print(f"• Resource ID: {RESOURCE_ID}")
        print(f"• Total de registres al fitxer: {result.get('total')}")
        print(f"• Columnes/Camps disponibles: {', '.join(fields)}\n")

        print(" Mostra dels registres (JSON):")
        for record in records[:5]:
            print(record)
    else:
        print("Error en obtenir les dades del recurs.")
else:
    print(f"Error en la petició datastore_search: Status {response.status_code}")
```
