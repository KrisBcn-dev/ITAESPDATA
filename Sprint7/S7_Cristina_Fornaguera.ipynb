### Función : validar_dades  
Recibe: variable dato y tipo de dato  
Acción : Validar el dato pasado por parámetro según su tipo (segundo parámetro) y, en caso de dato no válido, obliga su rectificación con un máximo de tres intentos  
Retorna : valor validado o False

```python
import re
import math

def validar_dades(dada, tipus):
  try:
    match tipus:
      case 'NomCognoms':
        if isinstance(dada, str) and len(dada.strip().split(' ')) >= 2:
          return dada.strip().capitalize()
      case 'Telefon':
        if re.match("^\\(?\\+?\\d{1,4}\\)?[-.\\s]?\\d{8,14}$",dada):
          return dada
      case 'AnysExp':
        if not isinstance(dada,int):
          dada = int(dada)
        if dada >= 0 and dada <= 70:
          return dada
      case 'Catala'|'Excel'|'Titol':
        if not isinstance(dada,int):
          dada = int(dada)
        if dada >= 0 and dada <= 3:
          return dada
      case _:
        return False
  except:
    return False
# test
print(validar_dades('7',"Catala"))
```

### Función: calcular_puntuacio  
Requisitos : Recibe el diccionario del candidato con los datos introducidos y validados  
Acción: Calcularemos la puntuación de la candidatura y la guardaremos en el mismo diccionario:  
puntuacio_total = experiencia (años * 1.5) + catalán + excel + titulación  
Retorna : Diccionario

```python
def calcular_puntuacio(candidat):
  candidat["puntuacio_total"] = float(candidat["AnysExp"]*1.5) + candidat["Catala"] + candidat["Excel"] + candidat["Titol"]
  return candidat

# test
calcular_puntuacio({'NomCognoms': 'Asd fd',
 'Telefon': '987654321',
 'AnysExp': 2.0,
 'Catala': 3,
 'Excel': 3,
 'Titol': 3})
```

### Función : classificar_candidat  
Requisitos : Recibe el diccionario del candidato con los datos introducidos y validados y su puntuación  
Acción: A partir de la puntuacio_total guardada en el diccionario se calcula la 'classificacio' que también se guarda en el diccionario de la candidatura siguiendo este criterio:  
+ classificacio = "Prioritat Alta" si la puntuacio_total es igual o mayor a 12
+ classificacio = "Candidat adequat" si la puntuacio_total es entre 8 y 11
+ classificacio = "No prioritari" si la puntuacio_total es menor a 8  

Retorna : Diccionario

```python
def classificar_candidat(candidat):
  if candidat["puntuacio_total"] >= 12:
    candidat["classificacio"] = "Prioritat Alta"
  elif candidat["puntuacio_total"] >= 8:
    candidat["classificacio"] = "Candidat adequat"
  else:
    candidat["classificacio"] = "No prioritari"
  return candidat

# test
classificar_candidat({'NomCognoms': 'Asd fd',
 'Telefon': '987654321',
 'AnysExp': 2.0,
 'Catala': 3,
 'Excel': 3,
 'Titol': 3,
 'puntuacio_total': 7.0})
```

### Función : main_seleccio # orquesta todas las funciones  
Requisitos : (Opcional) lista de candidaturas  
Acción :
+ Mostraremos un mensaje de funcionamiento del proceso para información de la persona candidata  
+ Definiremos la agrupación donde se guardarán los diccionarios de las candidaturas  
+ Crearemos un bucle de 10 iteraciones donde:  
+ - Se preguntará si se quiere introducir una candidatura (si es así seguimos)
+ - Crearemos un diccionario (candidatura)
+ - Utilizaremos las funciones anteriores en este orden y con el diccionario creado:  
+ - obtenir_dades -- Dentro llamará validar_dades a cada input
+ - calcular_puntuacio
+ - classificar_candidat
+ - mostrar_individual
+ Guardaremos el diccionario
+ Una vez terminado el bucle, llamaremos la función mostrar_informe y finalizará la ejecución del programa.

+ Gestión de errores: Permitiremos tres intentos por campo de cada candidatura,
y se permitirá la salida del programa si faltan candidatos y no se quieren o pueden añadir.

```python
def main_seleccio(candidatures = None):
  if candidatures is None: # sin datos iniciales
    candidatures = []
    print("Aquest programa et permetrà intoduïr les dades de 10 candidatures i en farà calculs")
    while len(candidatures)<10:
      candidatures.append(obtenir_dades())
      seguir = input("Vols afegir una altra candidatura? Si/No")
      if seguir.lower() == "no":
        break
  else:  # con datos iniciales
    # creamos diccionario de candidaturas a partir del listado con las claves iniciales definidas
    # para cada candidato y el listado pasado por argumento con las candidaturas
    claves = ["NomCognoms", "Telefon", "AnysExp", "Catala", "Excel", "Titol"]
    liscandidats = []
    num_campos = len(claves)  # En este caso, 6
    for i in range(0, len(candidatures), num_campos):
        # Extraemos el bloque de 6 elementos para este candidato
        valores_candidato = candidatures[i : i + num_campos]
        # Combinamos las claves con este bloque de valores
        candidato_dict = dict(zip(claves, valores_candidato))
        # Lo añadimos a la lista final
        liscandidats.append(candidato_dict)
    for c in liscandidats:
      for k in c:
        if validar_dades(c[k], k) is None: # si hay datos no validos en un listado proporcionado se eliminará la candidatura del grupo
          print("\nDades no vàlides", k, c[k])
          print("\nS'ha eliminat la candidatura següent:")
          print(c)
          liscandidats.remove(c)
          break
    # acotamos a un máximo de 10 candidatos, para más optimización,
    # este acotamiento también se podría realizar después de los cálculos y con los datos ordenados por mayor puntuación
    if len(liscandidats) > 10:
      liscandidats = liscandidats[:10]
  # una vez tenemos la lista validada procedemos a hacer los calculos y displays
  for can in liscandidats:
    mostrar_individual(classificar_candidat(calcular_puntuacio(can)))
  # mostramos el informe final
  mostrar_informe(liscandidats)
```

+ - Codificación : "Nombre|Edad|Síntomas (separados por comas)"

### Función : asignar_riesgo  
Argumentos : string  
Acción : Asignará un valor numérico según el criterio proporcionado a cada síntoma  
Retorno : int  

```python
def asignar_riesgo(sintoma):
    nivells_risc = {
        "fiebre_alta": 3,
        "dificultad_respirar": 5,
        "dolor_pecho": 4,
        "tos": 1,
        "fatiga": 1,
        "dolor_cabeza": 1
        }
    return nivells_risc.get(sintoma, 0)

# test
sintoma = "tos"
riesgo = asignar_riesgo(sintoma)
print(riesgo)
```

### Función : normalizar_datos  
Argumentos : String  
Acción : Crearemos un diccionario e introduciremos los datos de la cadena normalizados, y usaremos la función asignar_riesgo para cada síntoma, guardando los síntomas en un diccionario dentro del diccionario del paciente  
Retorno : diccionario de paciente

```python
def normalizar_datos(paciente):
    datos_paciente = {}
    datos = paciente.split("|")
    datos_paciente["nombre"] = datos[0]
    datos_paciente["edad"] = int(datos[1])
    sintomas = datos[2].split(",")
    sintomas_normalizados = [sintoma.strip().lower() for sintoma in sintomas]
    datos_paciente["sintomas"] = {}
    for sintoma in sintomas_normalizados:
        datos_paciente["sintomas"][sintoma] = asignar_riesgo(sintoma)
    return datos_paciente

# test
paciente = "Maria|45|tos, Fiebre Alta, dolor_pecho"
datos_paciente = normalizar_datos(paciente)
print(datos_paciente)
```

### Función : riesgo_total  
Argumentos : diccionario de paciente  
Acción : Suma los valores numéricos de los síntomas del paciente y los guarda en su mismo diccionario  
Retorno : diccionario de paciente

```python
def riesgo_total(paciente):
    riesgo = 0
    for sintoma, valor in paciente["sintomas"].items():
        riesgo += valor
    paciente["riesgo_total"] = riesgo
    return paciente

# test
paciente = normalizar_datos(paciente)
riesgo_total(paciente)
print(paciente)
```

### Función : ordena_prioridad  
Argumentos : diccionario de pacientes  
Acción : pasa el diccionario a dataframe y lo ordena de mayor riesgo_total a menor  
Retorno : dataframe

```python
import pandas as pd
def ordena_prioridad(pacientes):
    df = pd.DataFrame(pacientes)
    df = df.sort_values(by="riesgo_total", ascending=False)
    return df

#test
pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
pacientes_normalizados = [normalizar_datos(paciente) for paciente in pacientes]
pacientes_con_riesgo = [riesgo_total(paciente) for paciente in pacientes_normalizados]
df = ordena_prioridad(pacientes_con_riesgo)
display(df)
```

### Función : main_triaje  
Argumentos : Listado de pacientes  
Acción : Normaliza los datos de los pacientes, asignando valor numérico a sus síntomas, calcula su riesgo total y finalmente muestra una tabla con todos los datos, ordenados por el riesgo total calculado

+ Llamará a las funciones anteriores para generar una contraseña que cumpla con los argumentos pasados y validarla.

### Función :  generar_contrasenya  
Args : (Opcionales) longi int, may bool, min bool, num bool, sign bool  
Acción :  Genera una contraseña aleatoria con la longitud dada por param.,  
   conteniendo, o no, características alfanuméricas dadas por los demás   parámetros.  
   Si no se especifican argumentos se creará una contraseña segura estandar de 16 caracteres (todos True)
Retorno :  string

```python
import numpy as np
import string

# incializamos el generador de random
rng = np.random.default_rng()

def generar_contrasenya(longi = 16, may = True, min = True, num = True, sign = True):
  # generamos una pool de caracteres, numeros y signos
  pool = ''
  if may:
    pool += string.ascii_uppercase
  if min:
    pool += string.ascii_lowercase
  if num:
    pool += string.digits
  if sign:
    pool += string.punctuation

  # convertimos la pool en un array
  poolar = np.array(list(pool))

  # Usamos numpy para seleccionar los caracteres aleatoriamente
  selec = rng.choice(poolar, size=longi, replace=True)

  return "".join(selec)

# test
#generar_contrasenya(10, True, True, True, False)
generar_contrasenya()
```

### Función :  valida_contraseña  
Args :   pwd string, (Opcionales) : longitud int, may bool, min bool, num bool, sign bool  
Acción :  genera expresión regular según los argumentos recibidos y valida que la contraseña pwd cumpla con los criterios de los demás argumentos.  
Si no se especifican los argumentos opcionales, se verificará como unaa contraseña segura estandar de 16 caracteres (todos True)
Retorno :  bool. True si es correcta, False si no lo es

```python
import re

def valida_contraseña(pwd: str, longi = 16, may = True, min = True, num = True, sign = True):
  # Construimos la cadena de validación
  reg = '^'
  if may:
    reg += '(?=.*[A-Z])'
  if min:
    reg += '(?=.*[a-z])'
  if num:
    reg += '(?=.*[0-9])'
  if sign:
    reg += '(?=.*[^A-Za-z0-9])'

  # Construcción limpia del cuerpo de caracteres permitidos
  cuerpo = ''
  if may:
    cuerpo += 'A-Z'
  if min:
    cuerpo += 'a-z'
  if num:
    cuerpo += '0-9'
  if sign:
    # string.punctuation contiene: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
    # Escapamos los caracteres conflictivos para que Regex no se confunda
    simbolos = re.escape(string.punctuation)
    cuerpo += simbolos

  # Cerramos el patrón con la longitud exacta
  reg += '[' + cuerpo + ']{' + str(longi) + '}$'

  # retornamos la validación
  return bool(re.match(reg, pwd))

# test
valida_contraseña('asdf123$.', 9, False, True, True, True)
```

### Función :  crear_contrasenya  
Args :  longitud int, may bool, min bool, num bool, sign bool  
Acción :  generará contraseñas aleatorias siguiendo los argumentos de indicaciones, y validará dichas contraseñas, una a una hasta que la contraseña generada sea validada correctamente  
Retorno :  string contraseña generada aleatoriamente y validada

```python
def crear_contrasenya(longitud = 16, may = True, min = True, num = True, sign = True):
  contrasenya = generar_contrasenya(longitud, may, min, num, sign)
  while not valida_contraseña(contrasenya, longitud, may, min, num, sign):
    contrasenya = generar_contrasenya(longitud, may, min, num, sign)
  return contrasenya

# test
crear_contrasenya(10, 1, True, 15, 98)
```

### Función :  llegir_i_netejar_dades  
Args :  String. Ruta al fichero de texto (.txt) de los partidos  
Acción :  Abre el fichero pasado por argumento, lo lee a través de pandas indicando el separador, el nombre de las columnas y el motor (para evitar errores), y guarda en un dataframe.   
Estructura de línea original (.txt) : "Manlleu	0-1	Granollers"
+ Equipo local, goles local, Equipo visitante, goles visitante  
* Separación por tabulación '\t'  

Retorno :  dataframe de partidos

```python
import pandas as pd

def llegir_i_netejar_dades(docu):
  try:
    partits = {}
    # Leemos el fichero en formato columnas respetando los espacios complejos
    dfPartits = pd.read_csv(docu, sep=r'\t+|\s{2,}', header=None, names=['Local', 'Resultat', 'Visitant'], engine='python')
    dfPartits['Gols Local'] = dfPartits['Resultat'].str.split('-').str[0].astype(int)
    dfPartits['Gols Visitant'] = dfPartits['Resultat'].str.split('-').str[1].astype(int)
    dfPartits = dfPartits.drop('Resultat', axis=1)
  except FileNotFoundError as e:
    print(e)
    return None
  return dfPartits

# test
partits = llegir_i_netejar_dades('/content/drive/MyDrive/CURS ESP DATA/S7Data/historic partits.txt')
if partits is not None:
    print("Dades carregades correctament!")
    display(partits.head())
```

### Función :  calcular_estadistiques  
Args :  dataframe de partidos  
Acción :  extrae los datos necesarios del df y calcula por cada equipo:   
+ - el sumatorio de goles marcados  
+ - el sumatorio de goles recibidos
+ - los puntos ( siguiendo el criterio dado de "victoria + 3, empate +1, derrota +0"  
+ Guarda los datos en un diccionario de equipos  

Retorno : diccionario

```python
def calcular_estadistiques(partits):
  comptadors_equips = {}
  for index, row in partits.iterrows():
    local = row['Local']
    visitant = row['Visitant']
    gols_local = row['Gols Local']
    gols_visitant = row['Gols Visitant']
    if local not in comptadors_equips:
      comptadors_equips[local] = {'gols_favor': 0, 'gols_contra': 0, 'guanyats': 0, 'empatats': 0, 'perduts': 0, 'punts': 0}
    if visitant not in comptadors_equips:
      comptadors_equips[visitant] = {'gols_favor': 0, 'gols_contra': 0, 'guanyats': 0, 'empatats': 0, 'perduts': 0, 'punts': 0}

    # Acumulamos los goles marcados y recibidos de forma cruzada
    comptadors_equips[local]['gols_favor'] += gols_local
    comptadors_equips[local]['gols_contra'] += gols_visitant

    comptadors_equips[visitant]['gols_favor'] += gols_visitant
    comptadors_equips[visitant]['gols_contra'] += gols_local

    # Apliquem la lògica de punts de la lliga
    if gols_local > gols_visitant: # equipo local gana
      comptadors_equips[local]['guanyats'] += 1
      comptadors_equips[local]['punts'] += 3
      comptadors_equips[visitant]['perduts'] += 1
    elif gols_visitant > gols_local: # equipo local pierde
      comptadors_equips[local]['perduts'] += 1
      comptadors_equips[visitant]['punts'] += 3
      comptadors_equips[visitant]['guanyats'] += 1
    else: # empate
      comptadors_equips[local]['empatats'] += 1
      comptadors_equips[visitant]['empatats'] += 1
      comptadors_equips[local]['punts'] += 1
      comptadors_equips[visitant]['punts'] += 1

  return comptadors_equips

# test
comptadors_equips = calcular_estadistiques(partits)
display(comptadors_equips)
```

### Función :  generar_informe_final  
Args :  diccionario
Acción :  Convierte el diccionario en dataframe y respondemos a las preguntas planteadas:    
+ - Cuál es el equipo que ha marcado más goles?
+ - Cuál es el equipo que ha marcado menos goles?
+ - Número total de goles que ha marcado cada equipo  
+ - Tabla clasificatoria, equipos ordenados por puntos de mayor a menor.   
+ Muestra las preguntas y los resultados de cada una, con el informe final siendo la tabla clasificatoria  

Retorno :  None

```python
def generar_informe_final(comptadors_equips):
  df_est = pd.DataFrame.from_dict(comptadors_equips, orient='index')
  print("\n INFORME DE RESULTATS LLIGA CATALANA")
  print("\n- Equip més golejador: ", df_est['gols_favor'].idxmax())
  print("\n- Equip més golejat: ", df_est['gols_contra'].idxmax())
  print("\n- Equips de la lliga i gols marcats per cadascun :")
  display(df_est[['gols_favor']].sort_values(by='gols_favor', ascending=False))
  print("\n- Classificació final :")
  # Asignamos nombre al índice para que muestre la palabra Equip como en el ejemplo
  df_est.index.name = 'equip'
  display(df_est.sort_values(by='punts', ascending=False))
# test
generar_informe_final(comptadors_equips)
```
